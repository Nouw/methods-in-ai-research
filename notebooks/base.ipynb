```python
import pandas as pd


train_data = pd.read_csv(
    "../artifacts/splits/grouped/test_20260920-202201.csv",
    index_col="row_id",
    keep_default_na=False,
)


def show_examples(label: str, number: int = 25) -> None:
    examples = (
        train_data.loc[
            train_data["label"] == label,
            "utterance",
        ]
        .drop_duplicates()
        .sample(
            n=min(
                number,
                train_data.loc[
                    train_data["label"] == label,
                    "utterance",
                ].nunique(),
            ),
            random_state=42,
        )
    )

    for utterance in examples:
        print(f"- {utterance}")
```

```python
show_examples("ack")
```
