## Data processing

Set the variables:
- `data_path` the pathlike to the `.dat` file of the dataset
- `strategy` either original, grouped or both
- `output_path` pathlike where you want to store the splits

```python
from main import create_splits
from methods_in_ai_research.processing import load_dialog_acts

data_path = "../data/dailog_acts.dat"
strategy = "both"
output_path = "../artifacts/splits"

data = load_dialog_acts(data_path)

splits = create_splits(data, strategy, output_path)
```

## Training & Evaluation

You can run a model by using the run_pipeline method

```python
from main import run_pipeline

classifier = "bow-linear-svm"
split_name = "grouped"
train_data, test_data = splits[split_name]
training = True
evaluation = True
results_path = "../results"

run_pipeline(classifier_name=classifier, split_name=split_name, train_data=train_data, test_data=test_data, train_enabled=training, evaluate_enabled=evaluation, results_directory=results_path)
```

If you want to run all models (train and evaluate), just execute the following cell. You do have to set up the paths for the data and the splits.

```python
results_path = "../results"
split_name = "grouped"
train_data, test_data = splits[split_name]
training = True
evaluation = True

classifiers = ["rule-based", "bow-logistic-regression", "bow-linear-svm", "embedding-logistic-regression", "embedding-linear-svm"]

for classifier in classifiers:
    run_pipeline(classifier_name=classifier, split_name=split_name, train_data=train_data, test_data=test_data, train_enabled=training, evaluate_enabled=evaluation, results_directory=results_path)
```
